# matplot and seaborn

# scatterplot

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[60,70,80,90,50]
plt.scatter(x, y)# creates sclater plot using x and y value
plt.title("Scatter Plot") # initiate the title to the graph
plt.xlabel("study hrs") # labels the horizontal axis
plt.ylabel("score") # labels the vertical axis
plt.grid() # add the grid lines
plt.show()
```

```python
import matplotlib.pyplot as plt
#bar plot 
a = ["A", "B", "C", "D"]
b= [40, 55, 30, 45]
plt.bar(a, b ,color="Pink")#creates bar plot using a and b value
plt.title("Student in Sections")
plt.xlabel("Section")
plt.ylabel("Number of Students")
plt.show()
```

```python
#linechart 
import matplotlib.pyplot as plt
year1 = ["Jan", "Feb", "Mar", "Apr", "May"]
sales1 = [20, 35, 30, 45, 50]
year2 = ["Jan", "Feb", "Mar", "Apr", "May"]
sales2 = [15, 30, 30, 50, 60]

plt.plot(year1,sales1,color="pink" , label="year 1")
plt.plot(year2,sales2,color="purple",  label="year 2") # labeling the line of year 2 

plt.title("monthly sales of 1 year ")
plt.xlabel("Months")
plt.ylabel(" Sales")
plt.legend()# denotes the line at top left
plt.grid()
```

```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
desktop_users = [1200, 1500, 1400, 1800, 1600]
mobile_users = [800, 1100, 1300, 1700, 2100]

plt.plot(days, desktop_users, color="blue", label="Desktop")
plt.plot(days, mobile_users, color="orange", label="Mobile")

plt.title("Daily Website Visitors")
plt.xlabel("Days of the Week")
plt.ylabel("Number of Users")
plt.legend()
plt.grid()

plt.show()
```

# heat plot


```python
import seaborn as sns
import matplotlib.pyplot as plt

data = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
#heatmap with numbers inside
sns.heatmap(data)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
data = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
#cmaop="Blues" is used to change the color of the heatmap
sns.heatmap(data,cmap="Blues", annot=True)#annot=True is used to display the numbers inside the heatmap 
plt.show()
```

# Histogram

```python
import matplotlib.pyplot as plt
ages = [15, 18, 22, 25, 29, 31, 35, 42, 48, 55, 60]
plt.hist(
    ages,
    bins=5,
    color="orange",
    edgecolor="black"
)# creates histogram using ages value with 5 bins and orange color
plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Customer Count")
plt.grid(axis="y")
plt.show()
```
