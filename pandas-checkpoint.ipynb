1. d.loc[1] (Single Label)
What it does: Searches the index for the explicit label 1 and extracts that single row.

Return type: A Pandas Series where the column names (name, age, class) become the index of the output.

2. d.loc[[1, 3]] (List of Labels)
What it does: Searches the index for the explicit labels 1 and 3 and extracts both rows.

Return type: A filtered Pandas DataFrame keeping the original row-and-column layout.

```python
import pandas as pd
data ={"name":("aeson", "jeson", " keson"),
       "order":("momo", "fufu", "rice"),
          "rating":(4.5, 3.5, 2.5)}
d=pd.DataFrame(data,index=[1,2,3])
print(d.loc[[1,2]])#loc is used to access the data in the dataframe using the index and column name
print("......................................")
print(d.loc[1])
```

# Indexing 
1. iloc (Position-Based Selection)
iloc uses 0-based index numbers. Syntax pattern: df.iloc[row_position, column_position]

```python
import pandas as pd

df = pd.read_csv("data.csv")
first_row = df.iloc[0]# Get the first row (row index 0), all columns

single_cell = df.iloc[0, 1]# Get the value at row 0, column 1 (Name column)

slice_data = df.iloc[0:3, 0:3]# Slice rows 0 to 2 (excludes 3) and columns 0 to 2
print(first_row)
print("........................")
print(single_cell)
print("........................")
print(slice_data)
```
